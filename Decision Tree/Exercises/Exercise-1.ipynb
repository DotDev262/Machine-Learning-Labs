```python
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42) # Use numpy's seed for numpy functions

# --- Constants for Data Generation ---
N_SAMPLES = 10000
MIN_AGE = 20
MAX_AGE = 60
AGE_THRESHOLD = 30
LOW_INCOME = 0
HIGH_INCOME = 1
# Probability base rates
BASE_PROB_YOUNG_LOW_INCOME = 0.30
BASE_PROB_OLD_LOW_INCOME = 0.70
# Probability boosts for high income
HIGH_INCOME_BOOST_YOUNG = 0.05
HIGH_INCOME_BOOST_OLD = 0.10

# 1. Generate features using NumPy
# Generate 100 random ages between MIN_AGE and MAX_AGE
age = np.random.randint(MIN_AGE, MAX_AGE + 1, N_SAMPLES)

# Generate 100 random income levels (0 or 1)
income = np.random.choice([LOW_INCOME, HIGH_INCOME], N_SAMPLES)

# 2. Calculate the probability of buying a house (prob_buy)
# Initialize a probability array
prob_buy = np.zeros(N_SAMPLES)

# Apply rules using NumPy's vectorized operations (no explicit loop needed for this part)

# Condition 1: Age > 30 (Older Group)
is_old = age > AGE_THRESHOLD
prob_buy[is_old] = BASE_PROB_OLD_LOW_INCOME + (
    HIGH_INCOME_BOOST_OLD * (income[is_old] == HIGH_INCOME)
)

# Condition 2: Age <= 30 (Younger Group)
is_young = age <= AGE_THRESHOLD
prob_buy[is_young] = BASE_PROB_YOUNG_LOW_INCOME + (
    HIGH_INCOME_BOOST_YOUNG * (income[is_young] == HIGH_INCOME)
)

# 3. Determine 'buys_house' based on calculated probability
# Generate random floats between 0 and 1
random_chance = np.random.rand(N_SAMPLES)

# If the random chance is less than prob_buy, the person buys a house (1), otherwise not (0)
buys_house = (random_chance < prob_buy).astype(int)

# 4. Create DataFrame
df = pd.DataFrame({
    'age': age,
    'income': income,
    'buys_house': buys_house
})

print(df.head())
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics

# 1. Define Features (X) and Target (y)
X = df[['age', 'income']]
y = df['buys_house']

# 2. Split Data (for proper training and evaluation)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 3. Initialize and Train the Decision Tree Classifier
# We can set max_depth to a small number since the underlying logic is simple
dt_classifier = DecisionTreeClassifier(max_depth=3, random_state=42)
dt_classifier.fit(X_train, y_train)

# 4. Make Predictions
y_pred = dt_classifier.predict(X_test)

# 5. Evaluate the Model
accuracy = metrics.accuracy_score(y_test, y_pred)
print(f"Decision Tree Accuracy: {accuracy:.2f}")
```
