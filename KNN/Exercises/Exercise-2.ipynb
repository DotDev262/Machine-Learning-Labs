```python
import numpy as np
import pandas as pd

# 1. Define the existing data points (Brightness, Saturation)
data = np.array([
    [40, 20],
    [50, 50],
    [60, 90],
    [10, 25],
    [70, 70],
    [60, 10],
    [25, 80]
])

# Define the corresponding class labels
classes = np.array(['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue'])

# 2. Define the new entry
new_entry = np.array([45, 55])

# 3. Calculate Euclidean distance for each data point
distances = np.sqrt(np.sum(np.square(np.subtract(data, new_entry)), axis=1))

# 4. Create a DataFrame for a clean output
df = pd.DataFrame({
    'Brightness': data[:, 0],
    'Saturation': data[:, 1],
    'Class': classes,
    'Euclidean Distance': distances
})

print(f"--- New Entry: Brightness={new_entry[0]}, Saturation={new_entry[1]} ---\n")
print(df)
```
